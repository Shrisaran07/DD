```python
import py4hw
```

```python
class DetectorSequence011(py4hw.Logic):
    def __init__(self, parent, name, reset, a, r):
        super().__init__(parent, name)
        self.reset = self.addIn('reset', reset)
        self.a = self.addIn('a', a) 
        self.r = self.addOut('r', r) 
        self.state = 0
    def clock(self):
        if (self.reset.get()):
            self.state = 0 
            self.r.prepare(0)
        else:
            if (self.state == 0):
                self.r.prepare(0) 
                if (self.a.get() == 0):
                    self.state = 1
            elif (self.state == 1):
                self.r.prepare(0)
                if (self.a.get() == 1):
                    self.state = 2
            elif (self.state == 2):
                if (self.a.get() == 1):
                    self.r.prepare(1)
                    self.state = 3
                else:
                    self.r.prepare(0)
                    self.state = 1 
            elif (self.state == 3):
                self.r.prepare(0) 
                if (self.a.get() == 1):
                    self.state = 0
                else:
                    self.state = 1 
```
